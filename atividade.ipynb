## Criando o Banco de Dados SQLite
### empresa.db

```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("empresa.db")
cursor = conn.cursor()
```

### 1 - Criando as tabelas

```python
# Tabela 1: Funcionarios
cursor.execute("""
CREATE TABLE Funcionarios (
id INTEGER PRIMARY KEY,
nome TEXT,
cargo TEXT,
salario REAL,
data_contratacao TEXT not null
);
""")
```

```python
cursor.execute("Select * from Funcionarios")
pd.DataFrame(cursor.fetchall(), columns = ['id','nome','cargo','salario','data_contratacao'])
```

### 3 - Agregação e Ordenação

#### SOMA TOTAL DOS SALARIOS

```python
cursor.execute("Select SUM(salario) from funcionarios")
cursor.fetchone()[0]
```

#### AGRUPAMENTOS

* CALCULO MEDIO DO SALARIO POR CARGO

```python
cursor.execute("""
    SELECT cargo, AVG(salario) AS media
    FROM funcionarios
    GROUP BY cargo
""")

pd.DataFrame(cursor.fetchall(), columns=['cargo', 'media'])
```

* Quantidade de funcionarios por cargo

```python
cursor.execute("Select cargo, COUNT(*) as quantidade from funcionarios group by cargo")

pd.DataFrame(cursor.fetchall(), columns=['cargo', 'quantidade'])
```

### 4 - TRANSAÇÕES ROLLBACK

```python
# SIMULAÇÃO Transação 

try:
    conn.execute("BEGIN TRANSACTION") # Inicia a transação
    cursor.execute("INSERT INTO funcionarios VALUES (9, 'Carlos', 'Analista', 7000, '2026-05-01')") # inserindo 
    raise Exception("Erro simulado") # Simula um erro
    conn.commit() # Se tudo estiver certo, confirma a transação
except:
    conn.rollback() # Reverte em caso de erro
    print("Transação revertida por erro")
```

```python
# Transação verdadeira
try:
    conn.execute("BEGIN TRANSACTION") # Inicia a transação
    cursor.execute("INSERT INTO funcionarios VALUES (10, 'Eduardo', 'Chefe', 70000, '2026-09-29')") # inserindo 
    conn.commit() # Se tudo estiver certo, confirma a transação
except:
    conn.rollback() # Reverte em caso de erro
    print("Transação revertida por erro")
```

```python
cursor.execute("Select * from funcionarios order by salario desc")
pd.DataFrame(cursor.fetchall(), columns = ['id','nome','cargo','salario','data_contratacao'])
```

### CHAVE ESTRANGEIRA E JOINS

```python
# Adicionando a chave estrangeira em funcionarios
try:
    conn.execute("PRAGMA foreign_keys = ON")
    cursor.execute("ALTER TABLE funcionarios ADD COLUMN departamento_id INTEGER REFERENCES departamentos(id)")
except sqlite3.OperationalError:
    pass

# Relacionando cada funcionario a um departamento
cursor.executemany(
    "UPDATE funcionarios SET departamento_id = ? WHERE id = ?",
    [
        (2, 1),
        (3, 3),
        (1, 4),
        (1, 5),
        (2, 6),
        (2, 7),
        (3, 8),
        (4, 10),
    ]
)

conn.commit()
```

#### INNER JOIN

* Funcionarios e seus departamentos

```python
cursor.execute("""
    SELECT funcionarios.nome AS funcionario,
           departamentos.nome AS departamento
    FROM funcionarios
    INNER JOIN departamentos
        ON funcionarios.departamento_id = departamentos.id
""")

pd.DataFrame(cursor.fetchall(), columns=['funcionario', 'departamento'])
```

#### LEFT JOIN

* Criando as tabelas de projetos e da relação entre funcionarios e projetos

```python
cursor.execute("""
    CREATE TABLE projetos (
        id INTEGER PRIMARY KEY,
        nome_projeto TEXT NOT NULL
    )
""")

cursor.execute("""
    CREATE TABLE funcionarios_projetos (
        funcionario_id INTEGER,
        projeto_id INTEGER,
        PRIMARY KEY (funcionario_id, projeto_id),
        FOREIGN KEY (funcionario_id) REFERENCES funcionarios(id),
        FOREIGN KEY (projeto_id) REFERENCES projetos(id)
    )
""")

cursor.executemany(
    "INSERT INTO projetos VALUES (?, ?)",
    [
        (1, 'Sistema Financeiro'),
        (2, 'Aplicativo da Empresa'),
        (3, 'Dashboard de Indicadores'),
    ]
)

cursor.executemany(
    "INSERT INTO funcionarios_projetos VALUES (?, ?)",
    [
        (5, 1),
        (6, 2),
        (7, 2),
        (7, 3),
        (8, 3),
    ]
)

conn.commit()
```

```python
cursor.execute("""
    SELECT funcionarios.nome AS funcionario,
           projetos.nome_projeto AS projeto
    FROM funcionarios
    LEFT JOIN funcionarios_projetos
        ON funcionarios.id = funcionarios_projetos.funcionario_id
    LEFT JOIN projetos
        ON funcionarios_projetos.projeto_id = projetos.id
    ORDER BY funcionarios.nome
""")

pd.DataFrame(cursor.fetchall(), columns=['funcionario', 'projeto'])
```

### 6 - DCL (Controle de Dados)

- **DDL (Data Definition Language):** usado para definir e modificar a estrutura do banco de dados, como tabelas, colunas e restrições. Exemplos: `CREATE`, `ALTER` e `DROP`.
- **DML (Data Manipulation Language):** usado para inserir, alterar e excluir registros nas tabelas. Exemplos: `INSERT`, `UPDATE` e `DELETE`.
- **DQL (Data Query Language):** usado para consultar e recuperar dados armazenados. O principal comando é o `SELECT`.
- **DCL (Data Control Language):** usado para controlar o acesso e as permissões dos usuários sobre o banco de dados. Exemplos: `GRANT`, que concede permissões, e `REVOKE`, que remove permissões.

O DCL é importante para proteger os dados, permitindo que cada usuário tenha somente os acessos necessários para realizar suas funções.
