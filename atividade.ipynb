## Criando o Banco de Dados SQLite
### empresa.db

```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("empresa.db")
cursor = conn.cursor()
```

### 1 - Criando as tabelas

```python
# Tabela 1: Funcionarios
cursor.execute("""
CREATE TABLE Funcionarios (
id INTEGER PRIMARY KEY,
nome TEXT,
cargo TEXT,
salario REAL,
data_contratacao TEXT not null
);
""")
```

```python
cursor.execute("Select * from Funcionarios")
pd.DataFrame(cursor.fetchall(), columns = ['id','nome','cargo','salario','data_contratacao'])
```

### 2 - Inserindo os dados na tabela

```python
# Inserindo em funcionarios
cursor.executemany(
    "INSERT INTO Funcionarios VALUES (?, ?, ?, ?, ?)",
    [
        (5, 'Luna', 'Gerente', 12000, '2026-07-01'),
        (6, 'Joana', 'DEV', 4000, '2026-09-01'),
        (7, 'Mauricio', 'Analista', 5000, '2026-03-07'),
        (8, 'Giovanna', 'DEV', 6000, '2026-04-06'),
    ]
)

conn.commit()
```

```python
#vizualizndo
cursor.execute("Select * from Funcionarios")
pd.DataFrame(cursor.fetchall(), columns = ['id','nome','cargo','salario','data_contratacao'])
```

#### VIZUALIZANDO PELO SALARIO

```python
#Vizualizando dados por salario maior que 5000

cursor.execute("Select * from funcionarios where salario > 5000")
cursor.fetchall()
```

#### VIZUALIZANDO PELO CARGO


```python
#Vizualizando dados por cargo igual a gerente

cursor.execute("Select * from funcionarios where cargo = 'Gerente'")
pd.DataFrame(cursor.fetchall(), columns = ['id','nome','cargo','salario','data_contratacao'])
```

#### ATUALIZANDO DADOS



```python
# Dando aumento para um funcionario

cursor.execute("UPDATE funcionarios set salario = 10000 where id = 3")
conn.commit()
print("Registro atualizado com sucesso!")
```

```python
cursor.execute("Select * from funcionarios where id = 3")
cursor.fetchall()
```

#### EXCLUSÃO

```python
cursor.execute("DELETE FROM funcionarios where id = 2")
conn.commit()
print("Registro deletado com sucesso!")
```

```python
cursor.execute("Select * from funcionarios")
cursor.fetchall()
```

### 3 - Agregação e Ordenação

#### SOMA TOTAL DOS SALARIOS

```python
cursor.execute("Select SUM(salario) from funcionarios")
cursor.fetchone()[0]
```

#### MEDIA SALARIAL

```python
cursor.execute("Select AVG(salario) from funcionarios")
cursor.fetchone()[0]
```

#### MAIOR X MENOR

```python
cursor.execute("Select MAX(salario), MIN(salario) from funcionarios")
cursor.fetchone()
```
